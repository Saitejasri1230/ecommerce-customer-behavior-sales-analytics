```python
import pandas as pd

df = pd.read_excel("../data/raw/Online Retail.xlsx")

clean_df = df.copy()

clean_df = clean_df.drop_duplicates()

clean_df["InvoiceDate"] = pd.to_datetime(
    clean_df["InvoiceDate"],
    errors="coerce"
)

clean_df["IsCancellation"] = (
    clean_df["InvoiceNo"].astype(str).str.startswith("C")
)

clean_df["Revenue"] = (
    clean_df["Quantity"] * clean_df["UnitPrice"]
)

sales_df = clean_df[
    (~clean_df["IsCancellation"]) &
    (clean_df["Quantity"] > 0) &
    (clean_df["UnitPrice"] > 0) &
    clean_df["InvoiceNo"].notna() &
    clean_df["StockCode"].notna() &
    clean_df["InvoiceDate"].notna()
].copy()

sales_df.drop(columns=["IsCancellation"]).to_csv(
    "../data/processed/clean_sales.csv",
    index=False
)

print("Original records:", len(df))
print("Sales records:", len(sales_df))
```
